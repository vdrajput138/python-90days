# Below are research for dir()
Summary:

	•	`dir()`  helps inspect objects to see what attributes and methods they have.

	•	With no args, it shows locals in scope.

	•	With objects, modules, or classes, it shows their accessible features.

	•	Can be customized with `__dir__()`  in classes.


## test dir()

```python
a = 10
b = "hello"
print(dir())  # which includes 'a', 'b' & python built-in names..
```

## dir() - for list

```python
my_list = [1, 2, 3]
print(dir(my_list))
# We can perform below operations on list..

""" 
'append', 'clear', 'copy', 'count', 'extend', 'index', 'insert', 'pop', 'remove', 'reverse', 'sort' 
"""
```

## dir() - for user defined class

```python
class Person:
    def __init__(self, name):
        self.name = name

    def greet(self):
        print("Hello, " + self.name)


p = Person("Alice")
print(dir(p))
```

```python
class Custom:
    def __dir__(self):
        return ["custom_attr", "custom_method"]


c = Custom()
print(dir(c))
```
