##### 1. Implement custom abs() function.


```python
def custom_abs(x):
    if x < 0:
        return -x
    else:
        return x


num = float(input("Enter number to return absolute value of it: "))
print(f"Absolute value of {num} is ", custom_abs(num))
```

##### 2. Find all factors of a number.

```python
def find_factors(n):
    if n <= 0:
        ValueError("Number should be Positive Integer.")

    factors = []
    for i in range(1, n + 1):
        if n % i == 0:
            factors.append(i)
    return factors


fact = int(input("Enter number to get factors: "))
print(f"Factor of {fact}: ", find_factors(fact))
```

#### 3. Normalize a list of numbers to the scale 0-1 (formula).

```python
def normalize_0_1(values):
    min_v = min(values)
    max_v = max(values)
    if min_v == max_v:
        raise ValueError("all values in list are same.")
    return [(x - min_v) / (max_v - min_v) for x in values]


data = list(map(int, input("Enter number seperate with space: ").split()))
print("Data: ", normalize_0_1(data))
```

#### 4. Compute Euclidean distance between two points.

```python
###########
# For 2D
###########

import math


def ed_2D(a, b):
    x1, y1 = a
    x2, y2 = b
    return math.sqrt((x2 - x1) ** 2 + (y2 - y1) ** 2)


flag_a = True
while flag_a:
    a = list(map(int, input("Enter point_1 (x1, y1): ").split()))
    if len(a) != 2:
        raise ValueError("Enter only 2-Dimentional co-ordinates. ")
    else:
        flag_a = False

flag_b = True
while flag_b:
    b = list(map(int, input("Enter point_1 (x2, y2): ").split()))
    if len(b) != 2:
        raise ValueError("Enter only 2-Dimentional co-ordinates. ")
    else:
        flag_b = False

print("Euclidean Distance of two points: ", round(ed_2D(a, b), 2))

###########
# For 3D- working on it.
###########


def get_point(dim=3, prompt="Enter point (x, y, z): "):
    while True:
        s = input(prompt).strip()
        s = s.replace(",", " ")
        parts = s.split()
        if len(parts) != dim:
            print(f"Please enter exactly {dim} values separated by space or comma.")
            continue
        try:
            return [float(p) for p in parts]
        except ValueError:
            print("All coordinates must be numbers (e.g. 1, 2.5, -3).")
```

#### 5. Convert temperature series (Celsius → Kelvin).

```python
def farh_cel(n):
    return n * 9 / 5 + 32


temp_in_cel = int(input("Enter temp in cel: "))
temp_k = farh_cel(temp_in_cel)
print(f"Temp of {temp_in_cel} is ", temp_k)
```

#### 7. Extract digits from string "abc123def" using loops.

```python
ip_str = "abc4593def"
digit = ""

for char in ip_str:
    if char.isdigit():
        digit += char

print("Digits in string : ", digit)
```
