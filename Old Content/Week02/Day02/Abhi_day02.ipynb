##### 1. Implement custom abs() function.


```python
def custom_abs(x):
    if x < 0:
        return -x
    else:
        return x


num = float(input("Enter number to return absolute value of it: "))
print(f"Absolute value of {num} is ", custom_abs(num))
```

##### 2. Find all factors of a number.

```python
def find_factors(n):
    if n <= 0:
        ValueError("Number should be Positive Integer.")

    factors = []
    for i in range(1, n + 1):
        if n % i == 0:
            factors.append(i)
    return factors


fact = int(input("Enter number to get factors: "))
print(f"Factor of {fact}: ")
find_factors(fact)
```

#### 7. Extract digits from string "abc123def" using loops.

```python
ip_str = "abc4593def"
digit = ""

for char in ip_str:
    if char.isdigit():
        digit += char

print("Digits in string : ", digit)
```
