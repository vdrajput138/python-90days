```python
# 4
a = 4
b = 5
inp = input()
if inp == "+":
    c = 4 + 5
    print(c)
elif inp == "-":
    d = 4 - 5
    print(d)
elif inp == "*":
    e = 4 * 5
    print(e)
elif inp == "/":
    f = 4 / 5
    print(f)
elif inp == "//":
    g = 4 // 5
    print(g)
elif inp == "%":
    h = 4 % 5
    print(h)
```

```python
# 5 BMI calc
height = float(input("Enter you height in meter(m): "))
weight = int(input("Enter you Weight(kg): "))
BMI = round(weight / (height * height), 2)
print(f"your BMI is: {BMI}")
```
