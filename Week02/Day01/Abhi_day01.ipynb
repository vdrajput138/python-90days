#### 01. Create a list of numbers & compute mean manually.

```python
user_input = input("Enter values, seperated by comma(,): ")
numbers = list(map(float, user_input.split(",")))

total = 0

for num in numbers:
    total += num

mean = total / len(numbers)
print(f"Mean of numbers is {mean}")
```

#### 03. Write function is_prime(n).

```python
def is_prime(n):
    if n <= 1:
        return False
    for i in range(2, n):
        if n % i == 0:
            return False
    return True


number = int(input("Enter Number: "))

if is_prime(number):
    print(f"{number} is prime.")
else:
    print(f"{number} is Not prime.")
```
