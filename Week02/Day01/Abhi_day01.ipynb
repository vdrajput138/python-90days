#### 01. Create a list of numbers & compute mean manually.

```python
user_input = input("Enter values, seperated by comma(,): ")
numbers = list(map(float, user_input.split(",")))

total = 0

for num in numbers:
    total += num

mean = total / len(numbers)
print(f"Mean of numbers is {mean}")
```

#### 03. Write function is_prime(n).

```python
def is_prime(n):
    if n <= 1:
        return False
    for i in range(2, n):
        if n % i == 0:
            return False
    return True


number = int(input("Enter Number: "))

if is_prime(number):
    print(f"{number} is prime.")
else:
    print(f"{number} is Not prime.")
```

#### 05. Given list of numbers, compute variance manually.

```python
numb = [10, 20, 30, 40, 50]

# mean
total_sum = 0
for num in numb:
    total_sum += num
mean = total_sum / len(numb)
print(f"mean : {mean}")

# sq_diff and sum
sq_diff = 0
for num in numb:
    sq_diff += (num - mean) ** 2
print(f"Sq_diff : {sq_diff}")

# population variance
variance = sq_diff / len(numb)
simple_variance = sq_diff / (len(numb) - 1)

print("Variance:", variance)
print("Simple Variance:", simple_variance)
```

#### 9. Implement your own min() function.

```python
def my_min(a, b):
    if a < b:
        return a
    else:
        return b


num1 = float(input("Enter 1st Number: "))
num2 = float(input("Enter 2nd Number: "))
print("Minimum number from them is : ", my_min(num1, num2))
```

#### 10. Compute dot-product of two lists manually.

```python
def dot_product(a, b):
    if len(a) != len(b):
        raise ValueError("Both List lenth should be same to do Dot Product.")

    result = 0
    for x, y in zip(a, b):
        result += x * y

    return result


l1 = [1, 3, -5, 7]
l2 = [4, 5, 9, 3]

print("Dot Product: ", dot_product(l1, l2))
```
